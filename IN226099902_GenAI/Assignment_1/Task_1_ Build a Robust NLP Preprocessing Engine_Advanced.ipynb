Task 2:Build Advanced Preprocessing Function
Implement the following function:
def preprocess_text(text):
    pass

Required Features
Your function must include:
Remove numbers
Example: "I have 2 dogs" → "I have dogs"
Remove extra spaces
Example: "This is good" → "This is good"
Handle repeated characters
Example: "soooo goooood!!!" → "so good"
Convert text to lowercase
Example: "WOW!!! This is GREAT!!!" → "wow this is great"
Remove very short tokens
Remove words with length ≤ 2
Exception: keep meaningful words like "no", "not"
Remove URLs and email-like patterns
Example: "Visit http://example.com now" → "visit now"

```python
def preprocess_text(text):
    # Error handling
    if not text or text.strip() == "":
        return []

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # Remove emails
    text = re.sub(r"\S+@\S+", "", text)

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Handle repeated characters (soooo → so)
    text = re.sub(r"(.)\1{2,}", r"\1", text)

    # Remove special characters
    text = re.sub(r"[^a-z\s]", "", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenization
    tokens = text.split()

    # Remove short words (<=2 except 'no', 'not')
    tokens = [word for word in tokens if len(word) > 2 or word in ['no', 'not']]

    return tokens
```

Task 4: Token Analytics
For each processed sentence, compute:
Total number of tokens
Number of unique tokens
Average token length


```python
for s in sentences:
    tokens = preprocess_text(s)
    cleaned_sentence = " ".join(tokens)

    print("Original:", s)
    print("Tokens:", tokens)
    print("Cleaned:", cleaned_sentence)
    print("-" * 50)
```

```python
def token_analysis(tokens):
    total = len(tokens)
    unique = len(set(tokens))
    avg_len = sum(len(t) for t in tokens) / total if total > 0 else 0

    return total, unique, round(avg_len, 2)
```

```python
for s in sentences:
    tokens = preprocess_text(s)
    total, unique, avg = token_analysis(tokens)

    print("Sentence:", s)
    print(f"Total Tokens: {total}, Unique: {unique}, Avg Length: {avg}")
    print("-" * 50)
```

Task 5: Frequency Analysis
Combine all tokens and use:
from collections import Counter

Identify:
Top 10 most frequent words
Top 5 least frequent words


```python
all_tokens = []

for s in sentences:
    all_tokens.extend(preprocess_text(s))

counter = Counter(all_tokens)

print("Top 10 Frequent Words:")
print(counter.most_common(10))

print("\nLeast 5 Frequent Words:")
print(counter.most_common()[-5:])
```

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(" ".join(tokens))

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
```
