Problem 1: Social Media – Post Engagement Analyzer
Problem Statement
Create a function to analyze engagement on a social media post.

Rules
Calculate total likes
If Total Likes ≥ 1000 → Viral Post
Otherwise → Normal Engagement

Real-Time Use

Social media analytics platforms
Marketing dashboards

Hint
Use function
Use loop for calculations

Expected Output Format
Total Likes: 1050
Post Status: Viral Post


```python
def analyze_post_engagement():
    # Initialize total likes to 0
    total_likes = 0

    # Ask user how many posts to analyze
    n = int(input("Enter number of posts: "))

    # Loop to collect likes for each post
    for i in range(n):
        # Take likes input for each post
        likes = int(input(f"Enter likes for post {i+1}: "))
        
        # Add likes to total_likes
        total_likes += likes

    # Print total likes
    print("\nTotal Likes:", total_likes)

    # Check if post is viral or normal
    if total_likes >= 1000:
        print("Post Status: Viral Post")
    else:
        print("Post Status: Normal Engagement")


# Call the function
analyze_post_engagement()
```

Problem 4: Social Media – Duplicate Account Detection

Problem Statement
Create a function to detect duplicate usernames in a system.

Rules
If duplicates exist → Flag accounts
Else → Accounts are unique

Real-Time Use
Platform moderation
Fraud detection systems

Hint
Use function
Use set data structure

Expected Output Format
Duplicate Accounts Found: Yes


```python
def duplicate_account_detect(usernames):
    if len(usernames)!=len(set(usernames)):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

users=[]
n=int(input("Enter the number of users you want to enter the usernames: "))
for i in range(n):
    names=input(f"Enter the usernames{i+1}: ")
    users.append(names)

duplicate_account_detect(users)
```

Else → Regular Crop

Hint

Use list filtering logic

Expected Output Format

```python
def filter_premium_crops(prices):
    premium_crops = []
    # Using list filtering logic
    for price in prices:
        if price > 2000:
            premium_crops.append(price)
            
    print(f"Premium Crops: {premium_crops}")

# Taking list inputs from the user
crop_prices = []
num_crops = int(input("Enter the number of crops you want to check: "))

for i in range(num_crops):
    price = int(input(f"Enter price for crop {i+1}: "))
    crop_prices.append(price)

filter_premium_crops(crop_prices)
```

Problem 9: E-Learning – Student Attendance Eligibility System

Problem Statement
Create a function to calculate attendance percentage and check exam eligibility.

Rules
Attendance ≥ 75% → Eligible
Attendance < 75% → Not Eligible

Real-Time Use
Learning management systems
Internship tracking platforms

Hint
Use function
Use list and loop

Expected Output Format
Attendance Percentage: 80.0
Exam Eligibility: Eligible


```python
def check_attendance_eligibility(attendance_record):
    # Check if the list is empty
    if len(attendance_record) == 0:
        print("No attendance data provided.")
        return

    days_present = 0
    total_days = len(attendance_record)
    
    # Using loop to calculate total days present
    for day in attendance_record:
        days_present += day
        
    percentage = (days_present / total_days) * 100
    
    if percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"
        
    print(f"Attendance Percentage: {percentage}")
    print(f"Exam Eligibility: {status}")

# Taking attendance input from the user
student_attendance = []
total_classes = int(input("Enter the total number of classes: "))

for i in range(total_classes):
    # Asking user to enter 1 for present and 0 for absent
    status = int(input(f"Day {i+1} - Enter 1 for Present, 0 for Absent: "))
    student_attendance.append(status)

check_attendance_eligibility(student_attendance)
```

Problem 10: Smart Electricity Bill Analyzer

Problem Statement
You are building a smart electricity billing system for residential users.
Create a function that calculates the electricity bill and classifies usage level based on units consumed.

Billing Rules
First 100 units → ₹3 per unit
Next 100 units (101–200) → ₹5 per unit
Above 200 units → ₹7 per unit

Usage Classification
Total bill < ₹500 → Low Usage
₹500 – ₹1500 → Moderate Usage
₹1500 → High Usage

Real-Time Use
Smart meter systems
Electricity board billing software
Energy consumption analytics

Hint
Use function to calculate bill
Use conditional logic for slabs
Use return values (bill and usage status)


```python
def calculate_electricity_bill(units):
    bill = 0
    
    # Conditional logic for billing slabs
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)
        
    # Usage Classification
    if bill < 500:
        usage_status = "Low Usage"
    elif bill < 1500:
        usage_status = "Moderate Usage"
    else:
        usage_status = "High Usage"
        
    return bill, usage_status

# Taking input from the user
units_consumed = int(input("Enter the electricity units consumed: "))

# Function call and output
total_bill, usage_level = calculate_electricity_bill(units_consumed)

print(f"Units Consumed: {units_consumed}")
print(f"Total Bill: ₹{total_bill}")
print(f"Usage Level: {usage_level}")
```
