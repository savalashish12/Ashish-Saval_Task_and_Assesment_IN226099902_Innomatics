Problem 1: Mobile Recharge Validation System
Problem Statement
Create a function to validate a mobile recharge request.
Rules:
Recharge amount must be ≥ ₹50
Valid plans: 199, 299, 399, 599
User should be allowed to retry if invalid


```python
def validate_recharge():
    valid_plans=[199,299,399,599]
    while True:
        amount=int(input("Enter recharge amount: "))

        if amount>=50 and amount in valid_plans:
            print("Recharge Successful!!")
            break
        else:
            print("Invalid recharge. Please try again...")
            

validate_recharge()


# input 70,90, 200,199
```

Problem 2: Inventory Reorder Alert System
Problem Statement
Given a dictionary of products and stock quantities:
Print Reorder Alert if stock < 15


Print Stock OK otherwise

```python
def inventory_check(products):
    for product, stock in products.items():
        if stock<15:
            print(f"{product}: Reorder Alert")
        else:
            print(f"{product}: Stock OK")

products={
    "Laptop":20,
    "Mouse":10,
    "Keyboard":25,
    "Monitor":8
}
inventory_check(products)
```

Problem 3: Student Result Processing System
Problem Statement
Create a function that:
Accepts a list of student marks
Calculates average marks
Prints Pass if average ≥ 50 else Fail

```python
def result(marks_list):
    total=0
    for mark in marks_list:
        total+=mark
    average=total/len(marks_list)

    print("Average Marks:",average)

    if average>=50:
        return "Pass"
    else:
        return "Fail"
    
user_input=(input("Enter the marks(space seperated): "))
marks=list(map(int, user_input.split()))
final_result=result(marks)
print("Result: ",final_result)

# marks are 70 45 86 93 39 30 87
```

Problem 4: Cab Fare Estimator with Retry Option
Problem Statement
Create a cab fare calculator:
Base fare = ₹50
₹12 per km
Peak hour → 25% extra
Allow retry using while loop

```python
def fare_calculation():
    while True:
        try:
            distance=float(input("Enter the distance traveled in km: "))
            peak=input("Is it peak hour? (yes/no): ").lower()

            fare=50+(12*distance)

            if peak=="yes":
                fare+=fare*0.25

            print(f"Total Fare: ₹{fare}")
            break
        except:
            print("Invalid input. Please try again.")

fare_calculation()

# fare km 8 peak hour yes
```

Problem 5: Employee Attendance Eligibility Checker
Problem Statement
Create a function that:
Accepts attendance list ("P", "A")
Calculates attendance percentage
Returns Eligible if ≥ 75% else Not Eligible

Hint
Use loop to count present days
Use return statement

```python
def check_attendance(attendance):
    present=0
    for day in attendance:
        if day=="P":
            present+=1
        
    perecntage=(present/len(attendance))*100
    print("Attendance Percentage:",perecntage)

    if perecntage>=75:
        return "Eligible"
    else:
        return "Not Eligible"
attendance=["P","P","A","P","P"]
status=check_attendance(attendance)
print("Eligibility Status: ",status)
```

Problem 6: Password Strength Checker
Problem Statement
Create a function to check password strength.
Rules:
Minimum length = 8
Must contain at least one digit
Must contain at least one special character (@#$)

```python
def check_password(password):
    special_chars="@#$"

    has_digit=False
    has_special=False
    for char in password:
        if char.isdigit():
            has_digit=True
        if char in special_chars:
            has_special=True

    if len(password)>=8 and has_digit and has_special:
        return "Strong Password"
    else:
        return "weak Password"

password=input("Enter Password: ")
print(check_password(password))
```

```python
def check_password(password):
    special_chars = "@#$"

    has_digit = False
    has_special = False

    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_chars:
            has_special = True

    if len(password) < 8:
        return "Password must be at least 8 characters."

    if not has_digit:
        return "Password must contain at least one digit."

    if not has_special:
        return "Password must contain at least one special character (@#$)."

    return "Strong Password ✅"


# Login style retry system
attempts = 3

while attempts > 0:
    password = input("Create Password: ")
    result = check_password(password)

    if result == "Strong Password ✅":
        print("Password Created Successfully 🎉")
        break
    else:
        print("Error:", result)
        attempts -= 1
        print("Attempts left:", attempts)

if attempts == 0:
    print("Account Locked. Try again later.")
```
